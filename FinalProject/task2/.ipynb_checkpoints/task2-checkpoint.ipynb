### Preprocess images by resizing, normalising, and applying suitable augmentation

```python
# Define dataset path 
DATASET_PATH = "dataset"

# Define image size 
IMAGE_SIZE = (224, 224)

# Define batch size 
BATCH_SIZE = 32

# 1. Load Raw Datasets ONCE
raw_train_ds = tf.keras.utils.image_dataset_from_directory(
    directory=DATASET_PATH,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=True,
    validation_split=0.2,
    subset="training",
    seed=123
)

# FIXED: Renamed to raw_val_ds to match downstream references
raw_val_ds = tf.keras.utils.image_dataset_from_directory(
    directory=DATASET_PATH,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False,  # Deterministic for validation metrics
    validation_split=0.2,
    subset="validation",
    seed=123
)

class_names = raw_train_ds.class_names
num_classes = len(class_names)

# Print out all the classes
print("\nDiscovered", len(class_names), "classes:\n")
for i, j in enumerate(class_names):
    print(i + 1, "-", j)

# 2. Augmentation and Normalization Pipelines
train_augmentation_pipeline = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal_and_vertical"),
    tf.keras.layers.RandomRotation(0.1),
    tf.keras.layers.RandomBrightness(0.1),
    tf.keras.layers.Rescaling(1./255),
    tf.keras.layers.Normalization(
        mean=[0.485, 0.456, 0.406],
        variance=[0.229**2, 0.224**2, 0.225**2]
    )
])

test_normalization_pipeline = tf.keras.Sequential([
    tf.keras.layers.Rescaling(1./255),
    tf.keras.layers.Normalization(
        mean=[0.485, 0.456, 0.406],
        variance=[0.229**2, 0.224**2, 0.225**2]
    )
])

# FIXED: Explicit tf.keras.layers usage
cnn_augmentation_pipeline = tf.keras.Sequential([
    layers.RandomFlip("horizontal_and_vertical"),
    layers.RandomRotation(0.1),
])

# 3. Branch into Parallel Data Streams
# Branch A: Classical ML Datasets (Normalized)
train_ds_ml = raw_train_ds.map(
    lambda x, y: (train_augmentation_pipeline(x, training=True), y),
    num_parallel_calls=tf.data.AUTOTUNE
).prefetch(buffer_size=tf.data.AUTOTUNE)

test_ds_ml = raw_val_ds.map(
    lambda x, y: (test_normalization_pipeline(x, training=False), y),
    num_parallel_calls=tf.data.AUTOTUNE
).prefetch(buffer_size=tf.data.AUTOTUNE)

# Branch B: CNN Datasets (Raw images [0-255])
train_ds_cnn = raw_train_ds.map(
    lambda x, y: (cnn_augmentation_pipeline(x, training=True), y),
    num_parallel_calls=tf.data.AUTOTUNE
).prefetch(tf.data.AUTOTUNE)

test_ds_cnn = raw_val_ds.prefetch(tf.data.AUTOTUNE)

```

### Explore and visualise representative examples from different classes.

```python
# Get the number of classes
num_classes = len(class_names)

# init dictionary to store found classes 
found_classes = {}

# Loop through training batches 
for images, labels in train_ds_ml.as_numpy_iterator():
    # Loop through (image, class) in batch
    for img, lbl in zip(images, labels):
        # If the class hasn't been seen yet, add it too the dictionary 
        if lbl not in found_classes:
            found_classes[lbl] = img
        # If we've got an image from each class then stop 
        if len(found_classes) == num_classes:
            break
    # Same again for the outer loop 
    if len(found_classes) == num_classes:
        break

# define values for un-normalising 
mean = np.array([0.485, 0.456, 0.406])
std = np.array([0.229, 0.224, 0.225])

# define grid layout
cols = 4
rows = 4
plt.figure(figsize=(16, 4 * rows))

# Loop through the found images 
for idx, (lbl_idx, img) in enumerate(sorted(found_classes.items())):
    plt.subplot(rows, cols, idx + 1)
    
    # Un-normalise the image so they look normal 
    img_display = (img * std) + mean
    img_display = np.clip(img_display, 0.0, 1.0)

    # Display the image
    plt.imshow(img_display)
    plt.title(class_names[lbl_idx], fontsize=9, fontweight='bold')
    plt.axis("off")

plt.tight_layout()
plt.show()
```

### Extract suitable image features such as colour histograms, HOG, or texture features 

```python
def extractFeatures(img, mean, std):
    # De-normalize image back to uint8 (0-255)
    img_unnorm = (img * std) + mean
    img_unnorm = np.clip(img_unnorm, 0.0, 1.0)
    img_uint8 = (img_unnorm * 255).astype(np.uint8)
    
    # Convert to grayscale
    gray_img = cv2.cvtColor(img_uint8, cv2.COLOR_RGB2GRAY)
    
    # --- FEATURE 1: Color Histograms ---
    hist_r = cv2.calcHist([img_uint8], [0], None, [32], [0, 256]).flatten()
    hist_g = cv2.calcHist([img_uint8], [1], None, [32], [0, 256]).flatten()
    hist_b = cv2.calcHist([img_uint8], [2], None, [32], [0, 256]).flatten()
    color_features = np.concatenate([hist_r, hist_g, hist_b])
    
    # --- FEATURE 2: HOG (Shape) ---
    hog_features = hog(
        gray_img, 
        orientations=8, 
        pixels_per_cell=(16, 16),
        cells_per_block=(2, 2), 
        block_norm='L2-Hys', 
        visualize=False
    )
    
    # --- FEATURE 3: GLCM (Texture) ---
    # 32 gray levels for maximum matrix speedup
    gray_quantized = gray_img // 8
    glcm = graycomatrix(
        gray_quantized, 
        distances=[1, 3], 
        angles=[0, np.pi/4], 
        levels=32, 
        symmetric=True, 
        normed=True
    )
    
    contrast = graycoprops(glcm, 'contrast').flatten()
    correlation = graycoprops(glcm, 'correlation').flatten()
    energy = graycoprops(glcm, 'energy').flatten()
    homogeneity = graycoprops(glcm, 'homogeneity').flatten()
    
    texture_features = np.concatenate([contrast, correlation, energy, homogeneity])
    
    return np.concatenate([color_features, hog_features, texture_features])


def process_dataset_safe(dataset, mean, std, n_jobs=4):
    X_list = []
    y_list = []
    dataset = dataset.unbatch().batch(2048)
    
    # Streams dataset batch-by-batch (32 images at a time) to prevent RAM overload
    for images, labels in tqdm(dataset, desc="Processing Batches"):
        # Convert tensors to numpy once per batch
        imgs_np = images.numpy()
        lbls_np = labels.numpy()
        
        # Extract features in parallel for the current batch
        batch_features = Parallel(n_jobs=n_jobs, prefer="threads")(
            delayed(extractFeatures)(img, mean, std) 
            for img in imgs_np
        )
        
        X_list.extend(batch_features)
        y_list.extend(lbls_np)
        
    return np.array(X_list), np.array(y_list)

# Run feature extraction safely
print("Processing training data...")
X_train, y_train = process_dataset_safe(train_ds_ml, mean, std, n_jobs=4)

print("Processing testing data...")
X_test, y_test = process_dataset_safe(test_ds_ml, mean, std, n_jobs=4)

print(f"\nFinal Training Feature Matrix Shape: {X_train.shape}")
print(f"Final Testing Feature Matrix Shape: {X_test.shape}")
print(f"Final Testing Feature Matrix Shape: {X_test.shape}")
```

### Scale the data

```python
print("Scaling features...")
scaler = StandardScaler()

# Fit on training data and transform it
X_train_scaled = scaler.fit_transform(X_train_ml)

# Only transform the test data using the parameters learned from the training data
X_test_scaled = scaler.transform(X_test_ml)

print("Scaling complete.")
```

### Train and compare Classiscal ML models

```python
# Setup our models
models = {
    # LogisticRegression with 1000 iterations, allows time for the data to converge
    'Logistic Regression': SGDClassifier(loss='log_loss', max_iter=5000, random_state=42, n_jobs=-1),
    # Decision Tree Classifier with a max depth of 10 to prevent over-fitting
    'Decision Tree': DecisionTreeClassifier(max_depth=20, random_state=42),
    # Random Forest with 100 trees, more has show to not have much of an effect on accuracy
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
    'Gradient Boosting': HistGradientBoostingClassifier(random_state=42)
}

# Init results list
results = []

# Start the lopp through the models
for name, model in models.items():
    print("Training: ", name, "\n", flush=True)
    start = time.time()
    # Logist Regression requires the scaled data. The rest do not, I tried giving all models scaled data but found a considerable drop in performance. 
    if name == 'Logistic Regression':
        model.fit(X_train_scaled, y_train)
        trainingTime = time.time() - start 
        y_pred = model.predict(X_test_scaled)
    else:
        # Train the model on the training data
        model.fit(X_train, y_train)
        trainingTime = time.time() - start 
        # Create predictions from the model using the test data (remaining 20%)
        y_pred = model.predict(X_test)

    mins, secs = divmod(int(trainingTime), 60)
    
    # Generate Precision, Recall and F1-score data
    prec, rec, f1, _ = precision_recall_fscore_support(y_test, y_pred, average='macro')

    # Save to the results 
    results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, y_pred), # The % of correct predictions 
        'Precision (Macro)': prec, # True Positives / True Positves + False Positives 
        'Recall (Macro)': rec, # True Positves / True Positives + False Negatives 
        'F1-Score (Macro)': f1, # 2 * ((Precision * Recall) / (Precision + Recall)) - Harmonic Mean
        'Training Time': f"{mins}m {secs}s"
    })

# Format the results into a table, sort by Accuracy
results_df = pd.DataFrame(results).sort_values(by='Accuracy', ascending=False)
print("========================== Classical ML Model Comparison ==========================\n")
print(results_df.to_string(index=False))
```

```python


# 1. Build Model Architecture
base_model = EfficientNetB0(
    include_top=False,
    weights='imagenet',
    input_shape=(224, 224, 3)
)
base_model.trainable = False  # Freeze backbone initially

inputs = layers.Input(shape=(224, 224, 3))
x = preprocess_input(inputs)  # Scaled from raw [0, 255] pixels
x = base_model(x, training=False)  # Keep BatchNorm layers in inference mode
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)
outputs = layers.Dense(num_classes, activation='softmax')(x)

cnn_model = models.Model(inputs, outputs, name="EfficientNetB0_Transfer")

# 2. Phase 1: Train Top Classification Head Only
start = time.time()
cnn_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("--- Phase 1: Warmup Top Head (10 Epochs) ---")
history_phase1 = cnn_model.fit(
    train_ds_cnn,
    validation_data=test_ds_cnn,
    epochs=10
)

# 3. Phase 2: Fine-Tune Top 30 Backbone Layers
base_model.trainable = True

# Freeze all layers except the top 30
for layer in base_model.layers[:-30]:
    layer.trainable = False

# CRITICAL FIX: Keep Batch Normalization layers locked during fine-tuning
for layer in base_model.layers:
    if isinstance(layer, layers.BatchNormalization):
        layer.trainable = False

# Recompile with a lower learning rate
cnn_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

callbacks = [
    tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=4, restore_best_weights=True),
    tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=2)
]

print("\n--- Phase 2: Fine-Tuning Top 30 Backbone Layers ---")
history_phase2 = cnn_model.fit(
    train_ds_cnn,
    validation_data=test_ds_cnn,
    epochs=15,
    callbacks=callbacks
)

total_time = time.time() - start 
mins, secs = divmod(int(total_time), 60)
```

```python


# 1. Run inference on the validation set
print("\n--- Evaluating CNN Model ---")
y_true = []
y_pred_probs = []

for images, labels in test_ds_cnn:
    preds = cnn_model.predict(images, verbose=0)
    y_pred_probs.extend(preds)
    y_true.extend(labels.numpy())

y_true = np.array(y_true)
y_pred = np.argmax(np.array(y_pred_probs), axis=1)

# 2. Compute macro metrics
prec, rec, f1, _ = precision_recall_fscore_support(y_true, y_pred, average='macro', zero_division=0)
acc = accuracy_score(y_true, y_pred)

# 3. Add CNN entry to your results dataframe
cnn_results = {
    'Model': 'EfficientNetB0 (Fine-Tuned)',
    'Accuracy': acc,
    'Precision (Macro)': prec,
    'Recall (Macro)': rec,
    'F1-Score (Macro)': f1,
    'Training Time': f"{mins}m {secs}s",
}

# Append to your existing classical results dataframe
final_results_df = pd.concat([results_df, pd.DataFrame([cnn_results])], ignore_index=True)
final_results_df = final_results_df.sort_values(by='Accuracy', ascending=False)

print("\n========================== Final All-Model Comparison ==========================\n")
print(final_results_df.to_string(index=False))

# Save the trained model weights
cnn_model.save("efficientnet_b0_final.keras")
print("\nModel saved successfully as 'efficientnet_b0_final.keras'")
```
